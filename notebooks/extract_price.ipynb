```python
import pandas as pd

model_data_list = []
failed_urls = []
processed_urls = []
for idx, url in enumerate(italika_ok):
    print(f"[{idx+1}/{len(italika_ok)}] Procesando: {url}")
    response = generic_extractor.get_model_data(url)
    if response is not None:
        data_dict = response.model_dump() if hasattr(response, 'model_dump') else response.__dict__
        data_dict['url'] = url
        model_data_list.append(data_dict)
        print(f"  ✓ Información extraída correctamente.")
        processed_urls.append(url)
    else:
        print(f"  ✗ No se pudo extraer información. Se añadirá a la lista de pendientes.")
        failed_urls.append(url)
        # NO volvemos a scrapear, solo guardamos los nulls para revisar después
        model_data_list.append({
            "base_price": None,
            "net_price": None,
            "discount_amount": None,
            "model": None,
            "colors": None,
            "sku": None,
            "url": url
        })

print("\n--- URLs que fallaron y necesitan revisión manual o reintento: ---")
for u in failed_urls:
    print(u)
```
